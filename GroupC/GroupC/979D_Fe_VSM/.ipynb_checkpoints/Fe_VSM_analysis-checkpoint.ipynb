# VSM: Fe film (979D, Group C)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --- load -------------------------------------------------------------
# Row 0 of the file is the header (Field, Moment); row 1 is the units row (G, emu),
# so we skip it. utf-8-sig strips the BOM that would otherwise corrupt the "Field" name.
path = "979D_Fe_GroupC.csv"          # change to the full path if the file is elsewhere
df = pd.read_csv(path, encoding="utf-8-sig", skiprows=[1])
df = df.apply(pd.to_numeric, errors="coerce").dropna().reset_index(drop=True)

H = df["Field"].to_numpy()           # applied field, G (Oe)
M = df["Moment"].to_numpy()          # magnetic moment, emu

print(df.shape)
df.head()
```

```python
# --- split the sweep into legs ---------------------------------------
# The field goes 0 -> +Hmax (virgin curve) -> -Hmax -> +Hmax,
# so each change in the sign of dH/dt marks the start of a new leg.
step = np.sign(np.diff(H))
turn = np.where(step[1:] != step[:-1])[0] + 1          # indices of the turning points
edges = np.r_[0, turn + 1, len(H)]
legs = [slice(edges[i], edges[i + 1]) for i in range(len(edges) - 1)]

for i, s in enumerate(legs):
    print(f"leg {i}: points {s.start}-{s.stop-1}, H from {H[s.start]:.0f} to {H[s.stop-1]:.0f} G")
```

```python
# --- full hysteresis loop --------------------------------------------
labels = ["Virgin (0 \u2192 +H)", "+H \u2192 \u2212H", "\u2212H \u2192 +H"]
colors = ["tab:gray", "tab:red", "tab:blue"]

fig, ax = plt.subplots(figsize=(7, 5))
for s, lab, c in zip(legs, labels, colors):
    ax.plot(H[s], M[s] * 1e6, ".-", ms=3, lw=1, color=c, label=lab)
ax.axhline(0, color="k", lw=0.5); ax.axvline(0, color="k", lw=0.5)
ax.set_xlabel("Magnetic field, H (G)")
ax.set_ylabel("Moment, m (\u00d710$^{-6}$ emu)")
ax.set_title("Fe film: VSM hysteresis loop")
ax.legend(); ax.grid(alpha=0.3)
fig.tight_layout()
# fig.savefig("Fe_VSM_loop.png", dpi=300)
plt.show()
```

```python
# --- zoom on the low-field region (switching) -------------------------
fig, ax = plt.subplots(figsize=(7, 5))
for s, lab, c in zip(legs, labels, colors):
    ax.plot(H[s], M[s] * 1e6, ".-", ms=4, lw=1, color=c, label=lab)
ax.set_xlim(-1500, 1500)
ax.axhline(0, color="k", lw=0.5); ax.axvline(0, color="k", lw=0.5)
ax.set_xlabel("Magnetic field, H (G)")
ax.set_ylabel("Moment, m (\u00d710$^{-6}$ emu)")
ax.set_title("Low-field region")
ax.legend(); ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```

### Optional: remove the linear high-field background (substrate/holder)

```python
# The moment decreases slightly with field at high |H| (diamagnetic substrate/holder).
# Fit a straight line to the saturated regions on both sides and subtract it.
# Fit each polarity separately so the saturation moment itself does not bias the slope.
pos, neg = H > 5000, H < -5000
chi = np.mean([np.polyfit(H[pos], M[pos], 1)[0], np.polyfit(H[neg], M[neg], 1)[0]])
M_corr = M - chi * H
print(f"background slope = {chi:.3e} emu/G")

fig, ax = plt.subplots(figsize=(7, 5))
for s, lab, c in zip(legs, labels, colors):
    ax.plot(H[s], M_corr[s] * 1e6, ".-", ms=3, lw=1, color=c, label=lab)
ax.axhline(0, color="k", lw=0.5); ax.axvline(0, color="k", lw=0.5)
ax.set_xlabel("Magnetic field, H (G)")
ax.set_ylabel("Moment, m (\u00d710$^{-6}$ emu)")
ax.set_title("Fe film: loop after linear background subtraction")
ax.legend(); ax.grid(alpha=0.3)
fig.tight_layout()
plt.show()
```
