# VSM: Fe film on Si substrate
Loads the Out-of-plane and In-plane M–H data from the Excel file and plots the hysteresis loops.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Change this path if the notebook is in a different folder
FILE = "979D_Fe_GroupC.xlsx"

def load_vsm(path, sheet):
    # Row 1 = headers (Field, Moment); row 2 = units (G, emu) -> skip the units row
    df = pd.read_excel(path, sheet_name=sheet, skiprows=[1])
    df = df.apply(pd.to_numeric, errors="coerce").dropna()
    df.columns = ["H_G", "M_emu"]
    return df.reset_index(drop=True)

oop = load_vsm(FILE, "Out-of-plane")
ip  = load_vsm(FILE, "In-plane")

print("Out-of-plane:", oop.shape, "| In-plane:", ip.shape)
display(oop.head()); display(ip.head())
```

```python
fig, ax = plt.subplots(figsize=(7, 5), dpi=120)
ax.plot(ip["H_G"],  ip["M_emu"]  * 1e3, "o-", ms=2.5, lw=1, color="tab:red",  label="In-plane")
ax.plot(oop["H_G"], oop["M_emu"] * 1e3, "s-", ms=2.5, lw=1, color="tab:blue", label="Out-of-plane")
ax.axhline(0, color="k", lw=0.6); ax.axvline(0, color="k", lw=0.6)
ax.set_xlabel("Magnetic field, H (G)")
ax.set_ylabel("Magnetic moment, M (memu)")
ax.set_title("VSM M–H loops: Fe film on Si")
ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout()
plt.savefig("Fe_Si_VSM_MH.png", dpi=300)
plt.show()
```

```python
# Separate panels: each orientation on its own axes, plus a low-field zoom
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), dpi=120)

for ax, df, name, col in [(axes[0], oop, "Out-of-plane", "tab:blue"),
                          (axes[1], ip,  "In-plane",     "tab:red")]:
    ax.plot(df["H_G"], df["M_emu"] * 1e3, "-", color=col, lw=1.2)
    ax.axhline(0, color="k", lw=0.6); ax.axvline(0, color="k", lw=0.6)
    ax.set_title(name); ax.set_xlabel("H (G)"); ax.set_ylabel("M (memu)"); ax.grid(alpha=0.3)

# Low-field zoom to see coercivity / remanence
for df, name, col in [(ip, "In-plane", "tab:red"), (oop, "Out-of-plane", "tab:blue")]:
    m = df["H_G"].abs() <= 500
    axes[2].plot(df.loc[m, "H_G"], df.loc[m, "M_emu"] * 1e3, "-", color=col, lw=1.2, label=name)
axes[2].axhline(0, color="k", lw=0.6); axes[2].axvline(0, color="k", lw=0.6)
axes[2].set_title("Low-field zoom (|H| ≤ 500 G)")
axes[2].set_xlabel("H (G)"); axes[2].set_ylabel("M (memu)")
axes[2].legend(); axes[2].grid(alpha=0.3)

plt.tight_layout()
plt.savefig("Fe_Si_VSM_panels.png", dpi=300)
plt.show()
```

```python
# Optional: estimate coercivity (Hc), remanence (Mr) and max moment (Ms*) from the loop
def loop_params(df):
    H, M = df["H_G"].values, df["M_emu"].values
    # zero crossings of M(H) -> coercive field (average of |Hc| on both branches)
    idx = np.where(np.diff(np.sign(M)) != 0)[0]
    Hc = [H[i] - M[i] * (H[i+1] - H[i]) / (M[i+1] - M[i]) for i in idx]
    # zero crossings of H -> remanent moment
    jdx = np.where(np.diff(np.sign(H)) != 0)[0]
    Mr = [M[j] - H[j] * (M[j+1] - M[j]) / (H[j+1] - H[j]) for j in jdx]
    return dict(Hc_G=np.mean(np.abs(Hc)) if Hc else np.nan,
                Mr_emu=np.mean(np.abs(Mr)) if Mr else np.nan,
                Mmax_emu=np.max(np.abs(M)))

pd.DataFrame({"Out-of-plane": loop_params(oop), "In-plane": loop_params(ip)}).T
```
