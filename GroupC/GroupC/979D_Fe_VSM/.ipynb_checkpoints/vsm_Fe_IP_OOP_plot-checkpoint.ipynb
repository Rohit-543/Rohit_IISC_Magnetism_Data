# Fe film VSM: In-plane vs Out-of-plane
Plots the two separate CSV files (Field in G, Moment in emu). Put this notebook in the same folder as the CSV files.

```python
import numpy as np
import matplotlib.pyplot as plt

FILE_IP  = "979D_Fe_In-Plan.csv"       # in-plane file
FILE_OOP = "979D_Fe_out of plan.csv"   # out-of-plane file

def load_vsm(path):
    """Read Field, Moment columns; skips header/unit rows automatically."""
    H, M = [], []
    with open(path, encoding="utf-8-sig") as f:
        for line in f:
            parts = line.replace("\t", ",").replace(";", ",").strip().split(",")
            try:
                h, m = float(parts[0]), float(parts[1])
            except (ValueError, IndexError):
                continue                      # header / units / blank row
            H.append(h); M.append(m)
    return np.array(H), np.array(M)

H_ip,  M_ip  = load_vsm(FILE_IP)
H_oop, M_oop = load_vsm(FILE_OOP)
print(f"In-plane : {len(H_ip)} points, H = {H_ip.min():.0f} to {H_ip.max():.0f} G")
print(f"Out-plane: {len(H_oop)} points, H = {H_oop.min():.0f} to {H_oop.max():.0f} G")
```

```python
# Overlay: full range (top) and low-field zoom (bottom), like the slide
fig, ax = plt.subplots(2, 1, figsize=(6.5, 9))
for a, lim in zip(ax, (None, 500)):
    a.plot(H_oop, M_oop*1e4, "-s", color="k", ms=2, lw=0.6, label="OOP")
    a.plot(H_ip,  M_ip*1e4,  "-o", color="r", ms=2, lw=0.6, label="In-Plane")
    a.axhline(0, color="k", lw=0.8); a.axvline(0, color="k", lw=0.8)
    a.set_xlabel("Field (G)"); a.set_ylabel("Moment (emu) ×10$^{-4}$")
    a.legend(loc="lower right")
    if lim:
        a.set_xlim(-lim, lim)
        # auto y-range for the zoom window
        sel = lambda H, M: M[np.abs(H) <= lim]
        ym = max(np.abs(sel(H_ip, M_ip)).max(), np.abs(sel(H_oop, M_oop)).max())*1e4
        a.set_ylim(-1.15*ym, 1.15*ym)
plt.tight_layout()
plt.savefig("vsm_Fe_IP_vs_OOP.png", dpi=150)
plt.show()
```

```python
# Optional: each curve on its own panel
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))
ax[0].plot(H_ip,  M_ip*1e4,  "r-o", ms=2, lw=0.6); ax[0].set_title("In-plane")
ax[1].plot(H_oop, M_oop*1e4, "k-s", ms=2, lw=0.6); ax[1].set_title("Out-of-plane")
for a in ax:
    a.axhline(0, color="k", lw=0.8); a.axvline(0, color="k", lw=0.8)
    a.set_xlabel("Field (G)"); a.set_ylabel("Moment (emu) ×10$^{-4}$")
plt.tight_layout()
plt.show()
```
