# LSMO / STO (001) thin film: XRD analysis (Rigaku SmartLab `.ras`)

**Part A:** 2θ/ω scan: peak positions and out-of-plane lattice parameter.
**Part B:** φ scan: in-plane symmetry and epitaxy.

Put the two `.ras` files in the same folder as this notebook (or edit the paths below).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from scipy.optimize import curve_fit

file_2theta = "1.XRD_LSMO_Rocking_Curve.ras"   # 2theta/omega scan
file_phi    = "3_Phi_Scan_LSMO.ras"            # phi scan
```

## Helper functions

```python
def read_ras(path):
    # Return (header dict, DataFrame[angle, intensity, atten]) from a Rigaku .ras file
    header, data = {}, []
    in_data = False
    with open(path, "r", errors="ignore") as f:
        for line in f:
            line = line.strip()
            if line.startswith("*RAS_INT_START"):
                in_data = True
                continue
            if line.startswith("*RAS_INT_END"):
                in_data = False
                continue
            if in_data:
                parts = line.split()
                if len(parts) >= 2:
                    data.append([float(p) for p in parts[:3]])
            elif line.startswith("*") and " " in line:
                key, val = line[1:].split(" ", 1)
                header[key] = val.strip().strip('"')
    cols = ["angle", "intensity", "atten"][:len(data[0])]
    return header, pd.DataFrame(data, columns=cols)

def axis_position(header, name):
    # Value of a named goniometer axis (e.g. Chi, TwoTheta, Omega) from the header
    for k, v in header.items():
        if k.startswith("MEAS_COND_AXIS_NAME-") and v == name:
            return header.get("MEAS_COND_AXIS_POSITION-" + k.split("-")[1])

def pseudo_voigt(x, A, x0, w, eta, b):
    g = np.exp(-4*np.log(2)*((x - x0)/w)**2)
    l = 1/(1 + 4*((x - x0)/w)**2)
    return A*(eta*l + (1 - eta)*g) + b
```

# Part A: 2θ/ω scan and out-of-plane lattice parameter

```python
hdr1, df1 = read_ras(file_2theta)
x, y = df1["angle"].values, df1["intensity"].values
lam = float(hdr1.get("HW_XG_WAVE_LENGTH_ALPHA1", 1.540593))   # Cu Ka1, Angstrom

print(df1.shape)
for k in ["MEAS_SCAN_AXIS_X", "MEAS_SCAN_START", "MEAS_SCAN_STOP", "MEAS_SCAN_STEP",
          "MEAS_SCAN_SPEED", "MEAS_SCAN_SPEED_UNIT", "MEAS_COND_XG_VOLTAGE", "MEAS_COND_XG_CURRENT"]:
    print(f"{k:24s}: {hdr1.get(k)}")

# NOTE: despite the file name, this is a coupled 2theta/omega scan, not a rocking curve.
fig, ax = plt.subplots(figsize=(8, 5))
ax.semilogy(x, np.clip(y, 1, None), lw=1)
ax.set_xlabel("2θ/ω (deg)"); ax.set_ylabel("Intensity (counts)")
ax.set_title("LSMO / STO (001)")
ax.set_xlim(x.min(), x.max())
plt.tight_layout(); plt.savefig("lsmo_xrd.png", dpi=300); plt.show()
```

```python
# Peak search with parabolic refinement
ys = pd.Series(y).rolling(5, center=True, min_periods=1).mean().values
idx, _ = find_peaks(ys, prominence=0.0005*ys.max(), distance=20)
rows = []
for i in idx:
    sl = slice(max(i-2, 0), i+3)
    p = np.polyfit(x[sl], y[sl], 2)
    pos = -p[1]/(2*p[0]) if p[0] < 0 else x[i]
    rows.append({"2theta (deg)": round(pos, 4), "counts": round(y[i]),
                 "d (A)": round(lam/(2*np.sin(np.radians(pos/2))), 4)})
peaks_2t = pd.DataFrame(rows)
peaks_2t
```

```python
# Pseudo-Voigt fits of the STO (002) and LSMO (002) peaks -> c = 2*d002 = lambda/sin(theta)
def fit_peak(lo, hi, x0, A):
    m = (x > lo) & (x < hi)
    p, cov = curve_fit(pseudo_voigt, x[m], y[m], p0=[A, x0, 0.1, 0.5, 5],
                       bounds=([0, lo, 0.01, 0, 0], [1e8, hi, 2, 1, 1e4]))
    return p, np.sqrt(np.diag(cov))

results = []
for name, lo, hi, x0, A in [("STO (002)", 46.3, 46.7, 46.465, 1.85e6),
                            ("LSMO (002)", 46.85, 47.45, 47.11, 1e4)]:
    p, e = fit_peak(lo, hi, x0, A)
    d002 = lam/(2*np.sin(np.radians(p[1]/2)))
    results.append({"Peak": name, "2theta (deg)": round(p[1], 4), "err (deg)": round(e[1], 4),
                    "FWHM (deg)": round(p[2], 3), "d002 (A)": round(d002, 4),
                    "c = 2*d002 (A)": round(2*d002, 4)})
res = pd.DataFrame(results)
display(res)

c_film = res.loc[res.Peak == "LSMO (002)", "c = 2*d002 (A)"].iloc[0]
a_sto  = res.loc[res.Peak == "STO (002)",  "c = 2*d002 (A)"].iloc[0]
print(f"Out-of-plane lattice parameter of LSMO film: c = {c_film:.4f} A")
print(f"STO substrate check: a = {a_sto:.4f} A (literature 3.905 A)")

fig, ax = plt.subplots(figsize=(7, 4.5))
m = (x > 45.8) & (x < 48.2)
ax.semilogy(x[m], np.clip(y[m], 1, None), lw=1, label="data")
ax.axvline(res.iloc[1]["2theta (deg)"], color="r", ls="--", label=f"LSMO (002), c = {c_film:.3f} Å")
ax.axvline(res.iloc[0]["2theta (deg)"], color="k", ls="--", label="STO (002)")
ax.set_xlabel("2θ/ω (deg)"); ax.set_ylabel("Intensity (counts)"); ax.legend()
plt.tight_layout(); plt.show()
```

**Part A result:** c(LSMO) ≈ **3.855 Å** (2θ ≈ 47.11°). STO gives 3.906 Å, matching its known 3.905 Å, so no zero-offset correction is needed.
This is about 0.4-0.5 % below the bulk LSMO value (≈ 3.87-3.88 Å), consistent with in-plane tensile strain from the STO substrate.

# Part B: φ scan, in-plane symmetry and epitaxy

```python
hdr2, df2 = read_ras(file_phi)
phi, I = df2["angle"].values, df2["intensity"].values
tth, om, chi = (float(axis_position(hdr2, n)) for n in ["TwoTheta", "Omega", "Chi"])

print(df2.shape)
for k in ["MEAS_SCAN_AXIS_X", "MEAS_SCAN_START", "MEAS_SCAN_STOP", "MEAS_SCAN_STEP",
          "MEAS_SCAN_SPEED", "MEAS_SCAN_SPEED_UNIT"]:
    print(f"{k:24s}: {hdr2.get(k)}")
print(f"fixed 2theta = {tth} deg, omega = {om} deg, chi = {chi} deg")

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(phi, I, lw=0.8)
ax.set_xlabel("φ (deg)"); ax.set_ylabel("Intensity (counts)")
ax.set_title(f"φ scan: 2θ = {tth}°, χ = {chi}°")
ax.set_xlim(0, 360); ax.set_xticks(np.arange(0, 361, 45))
plt.tight_layout(); plt.savefig("lsmo_phi_scan.png", dpi=300); plt.show()
```

```python
# Locate peaks (1 deg smoothing), then fit each with a pseudo-Voigt in a +-3 deg window
ys = pd.Series(I).rolling(101, center=True, min_periods=1).mean().values
idx, _ = find_peaks(ys, prominence=0.05*ys.max(), distance=int(30/0.01))

rows = []
for i in idx:
    m = abs(phi - phi[i]) < 3
    p, cov = curve_fit(pseudo_voigt, phi[m], I[m], p0=[I[i], phi[i], 0.15, 0.5, 0],
                       bounds=([0, phi[i]-3, 0.01, 0, 0], [1e7, phi[i]+3, 3, 1, 1e4]))
    rows.append({"phi (deg)": p[1], "err (deg)": np.sqrt(np.diag(cov))[1],
                 "FWHM (deg)": p[2], "peak counts": I[m].max()})
phi_peaks = pd.DataFrame(rows).round(3)
phi_peaks["spacing to previous (deg)"] = phi_peaks["phi (deg)"].diff().round(3)
display(phi_peaks)

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(phi, I, lw=0.8)
for p in phi_peaks["phi (deg)"]:
    ax.axvline(p, color="r", ls=":", lw=0.8)
    ax.text(p, ax.get_ylim()[1]*0.97, f"{p:.1f}°", rotation=90, ha="right", va="top", fontsize=8)
ax.set_xlabel("φ (deg)"); ax.set_ylabel("Intensity (counts)")
ax.set_xlim(0, 360); ax.set_xticks(np.arange(0, 361, 45))
plt.tight_layout(); plt.savefig("lsmo_phi_peaks.png", dpi=300); plt.show()
```

```python
# Symmetry and epitaxy checks
n = len(phi_peaks)
pos = phi_peaks["phi (deg)"].values
spacing = np.diff(pos)
fold = 360.0/np.mean(np.r_[spacing, 360 - (pos[-1] - pos[0])])

# Any additional (e.g. rotated-domain) peaks? Mask +-5 deg around the main peaks.
mask = np.ones_like(phi, bool)
for p in pos:
    mask &= abs(phi - p) > 5
bg_mean, bg_max = I[mask].mean(), I[mask].max()
weakest = phi_peaks["peak counts"].min()

print(f"Number of peaks in 360 deg      : {n}  ->  {fold:.0f}-fold symmetry")
print(f"Peak spacings                   : {np.round(spacing, 2)} deg (ideal 90)")
print(f"Max deviation from 90 deg       : {np.max(np.abs(spacing - 90)):.2f} deg")
print(f"Peak FWHM                       : {phi_peaks['FWHM (deg)'].min():.2f} - {phi_peaks['FWHM (deg)'].max():.2f} deg")
print(f"Peak intensity spread           : {phi_peaks['peak counts'].min():.0f} - {phi_peaks['peak counts'].max():.0f} counts")
print(f"Background between peaks        : mean {bg_mean:.2f}, max {bg_max:.1f} counts")
print(f"Weakest peak / background max   : {weakest/bg_max:.1e}")
print(f"chi = {chi}° vs. angle between (001) and (111) = {np.degrees(np.arccos(1/np.sqrt(3))):.2f}° -> {{111}}-type reflection")
```

## Part B interpretation: epitaxy and in-plane symmetry

- **In-plane symmetry:** the φ scan at χ = 54.74° (the angle between (001) and {111}) shows exactly **four peaks, spaced by 90°** (89.7°, 90.2°, 90.3°). That is the **4-fold in-plane symmetry** expected for a cubic/pseudo-cubic (001)-oriented film. The slight deviations from 90° and the ±8 % intensity differences are typical of a small residual sample tilt or misalignment.
- **Single in-plane orientation (epitaxy):** only these four peaks appear. Between them the signal is at the noise floor (a few counts at most, against ≥ 10⁵ counts in the peaks), so there are **no rotated domains and no polycrystalline or randomly oriented fraction**. A film with random in-plane orientation would give a flat φ scan, and a 45°-rotated domain would add extra peaks. The film is therefore **epitaxial**, with one well-defined in-plane orientation.
- **Mosaic spread:** the fitted FWHM is only ~0.06-0.10° in φ, which indicates a very well-ordered in-plane arrangement. Instrumental broadening is included in these widths, so they are an upper limit on the true in-plane mosaic.
- **Caveat (cube-on-cube):** one φ scan at one 2θ cannot separate film and substrate. The usual relation for LSMO on STO (001) is cube-on-cube, LSMO(001)[100] ∥ STO(001)[100]. To confirm it, record φ scans at the film's own 2θ and at the substrate's 2θ and check that the peaks coincide in φ. A reciprocal-space map, for example around (103), would additionally show whether the film is fully strained to the substrate in-plane.
